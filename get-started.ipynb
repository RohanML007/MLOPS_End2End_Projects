## MLFlow Tracking Server

```python
#importing mlflow

import mlflow
```

```python
#Used in MLflow to set the tracking server URI where experiment metadata 
#(runs, parameters, metrics, etc.) will be logged

mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
mlflow.set_experiment("Check Localhost Connection")

with mlflow.start_run():
    mlflow.log_metric("test", 1)
    mlflow.log_metric("rohan", 2)
```
